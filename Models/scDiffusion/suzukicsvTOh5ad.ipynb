```python
import scanpy as sc
import numpy as np
import scipy.sparse as sp

# -----------------------
# paths
# -----------------------
in_path  = "data/suzuki.h5ad"
out_path = "data/suzukiSVD.h5ad"

# -----------------------
# load data
# -----------------------
adata = sc.read_h5ad(in_path)

X = adata.X
if sp.issparse(X):
    X = X.toarray()

# X shape: (N, D)
print("Original shape:", X.shape)

# -----------------------
# center data
# -----------------------
mu = X.mean(axis=0, keepdims=True)   # (1, D)
Xc = X - mu

# -----------------------
# FULL SVD
# -----------------------
U, S, Vt = np.linalg.svd(Xc, full_matrices=False)
# U: (N, r), S: (r,), Vt: (r, D)
# r = min(N, D)

Z = U * S    # (N, r)

print("Latent shape (Z):", Z.shape)

# -----------------------
# build new AnnData
# -----------------------
adata_svd = sc.AnnData(
    X=Z.astype(np.float32),
    obs=adata.obs.copy()
)

# store decoding info
adata_svd.uns["svd_mean"] = mu.astype(np.float32)
adata_svd.uns["svd_Vt"]   = Vt.astype(np.float32)
adata_svd.uns["svd_S"]    = S.astype(np.float32)

# keep reference to genes
adata_svd.var_names = [f"SVD_{i}" for i in range(Z.shape[1])]

# -----------------------
# save
# -----------------------
adata_svd.write(out_path)

print("Saved:", out_path)
```
