```python
import numpy as np

# expr_T is a pandas DataFrame of shape (10, 22206)

# 1) DataFrame -> numpy (numeric only)
X = expr_T.select_dtypes(include=[np.number]).to_numpy(dtype=np.float32)
print("X shape:", X.shape)

# 2) (recommended) z-score normalize per gene/feature
X = (X - X.mean(axis=0, keepdims=True)) / (X.std(axis=0, keepdims=True) + 1e-6)

# 3) Full SVD
U, S, VT = np.linalg.svd(X, full_matrices=False)
print("U:", U.shape, "S:", S.shape, "VT:", VT.shape)

# 4) Sample embeddings (latent): Z = U * S  (same as U @ diag(S))
Z = U * S
print("Z shape (full rank):", Z.shape)  # (10, 10)

pd.DataFrame(Z).to_csv("data/suzukiSVDcsv.csv", index=False)
```
